```python
expression = "1 + 2"
```

```python
expression = expression.replace(' ' ,'')
```

```python
stack = []
```

```python
for c in expression:
    stack.append(c)
stack
```

```python
for i in range(len(expression)):
    print(i)
```

```python
def calc(op1: int, op2: int, oper: str):
    if oper == "+":
        return op1 + op2
    elif oper == "-":
        return op1 - op2
        
```

```python
while len(stack) > 0:
    op2 = stack.pop()
    oper = stack.pop()
    op1 = stack.pop()
```

```python
calc(int(op1), int(op2), oper)
```
