```python
from collections import Counter
```

```python
# inputs = ["eat", "tea", "tan", "ate", "nat", "bat"]
inputs = ["eat", "tea", "tan", "ate", "nat", "bat", "Eat"]
```

```python
counters = [Counter(i) for i in inputs]
counters
```

```python
counters[0] == counters[-1]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(counters, index=inputs).fillna(0).astype(int)
df
```

```python
list(df.columns)
```

```python
df.groupby(
    list(df.columns) # 'e', 'a', 't', 'n', 'b'
).apply(lambda g: g.index.tolist()).tolist()
```
