```python
from typing import NamedTuple
import dataclasses
```

```python
@dataclasses.dataclass
class Point:
    x: int
    y: int

@dataclasses.dataclass
class Metadata:
    name: str
    description: str
```

```python
d = {}
d['point1'] = (Point(x=1, y=2), Metadata(name='point1', description='This is point 1'))
d['point2'] = (Point(x=3, y=4), Metadata(name='point2', description='This is point 2'))
d
```

```python
d.get('point1')
```

```python
point_with_metadata = NamedTuple('point_with_metadata', [('point', Point), ('metadata', Metadata)])
d = {}
d['point1'] = point_with_metadata(point=Point(x=1, y=2), metadata=Metadata(name='point1', description='This is point 1'))
d['point2'] = point_with_metadata(point=Point(x=3, y=4), metadata=Metadata(name='point2', description='This is point 2'))
d
```

```python
d.get('point1').metadata
```
