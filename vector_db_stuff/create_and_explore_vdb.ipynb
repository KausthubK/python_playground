````python
import chromadb
````

````python
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "A cat naps peacefully in the sun.",
    "Dogs are known for their loyalty and playful nature.",
    "The sun rises in the east and sets in the west.",
    "Japan is the land of the rising sun.",
    "The cat sat on the mat.",
    "Whisky is a specialty of Japan, Scotland, Ireland, and USA."
]
````

### 2. Initialization of the Vector Database Client and Collection:

After installation, initialize the client for your chosen vector database and create or retrieve a collection. A collection in a vector database is analogous to a table in a relational database, storing a set of vectors and their associated metadata.

````python
client = chromadb.Client()
collection_name = "example_docs"
collection = client.get_or_create_collection(name=collection_name)
````

### 3. Adding Data and Generating Embeddings:

Add your data (e.g., text documents) to the collection. The vector database will automatically handle the conversion of this data into numerical vector embeddings, often using a pre-trained model like a sentence transformer. You also typically provide unique IDs for each document.

````python
collection.add(documents=documents, ids=[f"doc{i}" for i in range(len(documents))])
````

### 4. Querying for Similar Data:

Perform similarity searches by providing a query (e.g., a text string) to the collection. The vector database will convert your query into an embedding and then find the most semantically similar documents based on the distance between their embeddings. You can specify the number of results to retrieve.

````python
results = collection.query(query_texts=["animals that are playful"], n_results=2)
results
````

````python
results = collection.query(query_texts=["animals that love the sun"], n_results=2)
results
````
