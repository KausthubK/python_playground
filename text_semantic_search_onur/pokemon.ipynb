```python
import os
import dotenv
dotenv.load_dotenv(dotenv_path='.env')

from openai import OpenAI

from dataset import Pokedex
from embeddings import generate_embeddings
from vector_store import VectorStore
```

```python
client = OpenAI(api_key=os.environ['OPENAI_API_KEY'])
```

```python
dex = Pokedex()
```

```python
embeddings = generate_embeddings(texts=dex.df.description.tolist(), oai_client=client)
```

```python
len(embeddings[0])
```

```python
vs = VectorStore(dimension=len(embeddings[0]))
```

```python
vs.add(emb=embeddings, ids=dex.df.national_number.tolist())
```

```python
query = "legendary fire"
```

```python
query_emb = generate_embeddings(texts=[query], oai_client=client)[0]
```

```python
d, i = vs.search(query_embedding=query_emb)
```
